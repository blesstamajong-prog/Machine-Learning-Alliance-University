```python
df["Sex"].value_counts().plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.title("Patients by Gender")
plt.show()
```

```python
df["Drug"].value_counts().idxmax() 
```

```python
df["Drug"].value_counts().plot(
kind="pie",
autopct="%1.1f%%"
)
plt.title("Proportion of Each Drug")
plt.ylabel("")
plt.show()
```

```python
plt.hist(df["Age"])
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.title("Age Distribution of Patients")
plt.show()
```

```python
df["Sex"].value_counts().plot(
kind="pie",
autopct="%1.1f%%"
)
plt.title("Male and Female Patient Distribution")
plt.ylabel("")
plt.show()
```

```python
sns.countplot(x="BP", hue="Drug", data=df)
plt.xlabel("Blood Pressure Level")
plt.ylabel("Number of Patients")
plt.title("Drug Prescription Across Different BP Levels")
plt.show()
```

```python
sns.boxplot(x="Drug", y="Age", data=df)
plt.xlabel("Drug Type")
plt.ylabel("Age")
plt.title("Age Distribution Across Different Drug Types")
plt.show()
```

```python
plt.scatter(df["Age"], df["Na_to_K"])
plt.xlabel("Age")
plt.ylabel("Na to K Ratio")
plt.title("Relationship Between Age and Na to K Ratio")
plt.show()
```
