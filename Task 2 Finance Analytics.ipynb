```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
df = pd.read_csv("2. Finance_Analytics_Dataset.csv")
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
print(df["PaymentMode"].value_counts())
```

```python
print(df["Category"].value_counts())
```

```python
sns.countplot(data=df, x="PaymentMode")
plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
sns.countplot(data=df[df["PaymentMode"].isin(["Debit Card", "Credit Card"])], x="PaymentMode")
plt.title("Debit vs Credit Card Amount Comparison")
plt.show()
```

```python
sns.barplot(data=df, x="PaymentMode", y="Amount")
```

```python
sns.barplot(data=df, x="CustomerSegment", y="Amount")
plt.title("Spending pattern By Customer")
```

```python
df["PaymentMode"].value_counts().plot.pie()
plt.title("payment Mode Distribution")
```

```python
sns.boxplot(data=df, x="Category", y="Amount")
plt.title("Tranasaction Amount by Category")
```
